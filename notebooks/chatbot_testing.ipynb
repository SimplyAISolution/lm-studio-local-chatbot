# LM Studio Local Chatbot - Testing Notebook

This notebook demonstrates how to use the LM Studio chatbot API for interactive testing and experimentation.

## Setup

First, let's import the necessary modules and initialize the chatbot.

```python
import sys
from pathlib import Path

# Add src to path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from chatbot import LMStudioChatbot
from rich.console import Console

console = Console()
```

## Initialize Chatbot

Create a chatbot instance with the default configuration.

```python
# Initialize chatbot
chatbot = LMStudioChatbot()
print("Chatbot initialized successfully!")
```

## List Available Models

Check what models are available in your LM Studio instance.

```python
# List available models
models = chatbot.list_models()
print(f"\nFound {len(models)} models:")
for i, model in enumerate(models, 1):
    print(f"{i}. {model}")
```

## Load a Model

Load a specific model to use for chat. Replace the model name with one from the list above.

```python
# Load a model (replace with your model name)
model_name = models[0] if models else "local-model"
chatbot.load_model(model_name)
print(f"Loaded model: {model_name}")
```

## Send a Message

Test the chatbot by sending a message.

```python
# Send a test message
response = chatbot.chat("Hello! Can you introduce yourself?")
if response:
    print("\nAssistant response:")
    print(response)
```

## Continue the Conversation

The chatbot maintains conversation history automatically.

```python
# Continue the conversation
response = chatbot.chat("What can you help me with?")
if response:
    print("\nAssistant response:")
    print(response)
```

## View Conversation History

Check the conversation history.

```python
import json

print("\nConversation History:")
print(json.dumps(chatbot.conversation_history, indent=2))
```

## Save Conversation

Save the conversation to a file for later reference.

```python
# Save conversation
chatbot.save_conversation("../logs/notebook_conversation.json")
print("Conversation saved!")
```

## Clear History

Clear the conversation history to start fresh.

```python
# Clear history
chatbot.clear_history()
print("History cleared!")
```

## Experiment with Parameters

Try different temperature and max_tokens settings.

```python
# Adjust parameters
chatbot.temperature = 0.9  # More creative
chatbot.max_tokens = 500   # Shorter responses

response = chatbot.chat("Tell me a creative story about a robot.")
if response:
    print("\nCreative response:")
    print(response)
```

## Testing Different Prompts

Test various prompts to evaluate model performance.

```python
test_prompts = [
    "Explain quantum computing in simple terms.",
    "Write a haiku about programming.",
    "What are the benefits of running LLMs locally?"
]

for prompt in test_prompts:
    print(f"\n{'='*60}")
    print(f"Prompt: {prompt}")
    print('='*60)
    
    response = chatbot.chat(prompt)
    if response:
        print(response)
    
    # Clear history between tests
    chatbot.clear_history()
```
